# Getting Started with idris-ml

This notebook introduces the idris-ml library interactively. All library modules are
pre-loaded via `Notebook.Prelude` — no imports needed.

The kernel wraps the Idris 2 REPL, so you can use all standard REPL commands:
`:t`, `:doc`, `:browse`, `:module`, `:exec`, and `:let`.

## Type Queries

Use `:t` to inspect types — the core of type-driven development.

```idris2
:t Var
```

```idris2
:t linearLayer
```

```idris2
:t MkDataPoint
```

## Documentation and Browsing

```idris2
:doc linearLayer
```

```idris2
:browse Math
```

## Expressions

Pure expressions are evaluated directly.

```idris2
2 + 3
```

```idris2
(\x => x * x) 7
```

```idris2
the (Vect 3 Nat) [1, 2, 3]
```

## Definitions

Type signatures and function clauses are auto-prefixed with `:let`.
Definitions persist across cells.

```idris2
square : Double -> Double
square x = x * x
```

```idris2
:exec putStrLn (show (square 7.0))
```

```idris2
:t square
```

## IO with `:exec`

Side-effecting code needs `:exec`. Use `let...in` chains for bindings.

```idris2
:exec putStrLn "Hello from idris-ml!"
```

```idris2
:exec do { srand 42;
  r <- normalSample;
  putStrLn ("Normal sample: " ++ show r) }
```

## Module System

All modules are pre-loaded, but you can import additional ones with `:module`.
Imports persist across cells.

```idris2
:module Layer.Core
```

```idris2
:browse Layer.Core
```
